```python

class Accessoire(list):
    pass



class Pic(Accessoire):

    def embrocher(self, postit):
        self.append(postit)

    def liberer(self):
        if len(self) > 0:
            postit = self.pop()
            return postit
        return None

class Bar(Accessoire):

    def recevoir(self, commande):
        self.append(commande)

    def evacuer(self):
        if len(self) > 0:
            commande = self.pop()
            return commande
        return None


```

```python
import sys
import time
import re


def usage():
    print(f"usage: {sys.argv[0]} fichier")
    exit(1)


class Clients:
    def __init__(self, fname):
        commandes = []
        start = time.time()
        fmt = re.compile(r"(\d+)\s+(.*)")

        with open(fname, "r") as f:
            for line in f:
                found = fmt.search(line)

                if found:
                    when = int(found.group(1))
                    what = found.group(2)
                    commandes.append((start + when, what.split(",")))

        self.commandes = commandes[::-1]

    def commande(self):
        if len(self.commandes) > 0:
            while True:
                if time.time() > self.commandes[-1][0]:
                    return self.commandes.pop()[1]
        else:
            return None
```

```python
class Employe:
    def __init__(self, pic, bar, clients):
        self.pic = pic
        self.bar = bar
        self.clients = clients


class Serveur(Employe):

    def prendre_commande(self):
        commande = self.clients.commande()

        if commande is None:
            return False

        print("Serveur: prêt pour prendre une nouvelle commande...")
        print(f"Serveur: j'ai la commande '{commande}'")
        print(f"Serveur: j'écris sur le post-it '{commande}'")

        self.pic.embrocher(commande)
        return True

    def servir(self):
        commande = self.bar.evacuer()

        if commande is not None:
            print(f"Serveur: j'apporte la commande '{commande}'")
            for cocktail in commande:
                print(f"Serveur: je sers '{cocktail}'")
            return True

        return False

    def run(self):
        while self.prendre_commande():
            pass
```

```python
class Bariste(Employe):

    def preparer(self):
        commande = self.pic.liberer()

        if commande is not None:
            print(f"Bariste: je commence la fabrication de '{commande}'")

            for cocktail in commande:
                print(f"Bariste: je prépare '{cocktail}'")

            self.bar.recevoir(commande)

            print(f"Bariste: la commande {commande} est prête")
            return commande

        return None

    def run(self):
        while True:
            commande = self.preparer()

            if commande is None:
                break
```

```python
def main():
    le_pic = Pic()
    le_bar = Bar()

    les_clients = Clients("lundi")

    alice = Serveur(le_pic, le_bar, les_clients)
    bob = Bariste(le_pic, le_bar, les_clients)

    alice.run()
    bob.run()

    while alice.servir():
        pass


if __name__ == "__main__":
    main()
```
